### 2 convert temperature 

```python
def convert_temp(temp, temp__unit):
    if(temp__unit=='F'):
        return temp*9/5 - 32
    elif temp__unit=='C':
        return (temp-32)*5/9
    else: 
        return None
```

```python
print(convert_temp(32,'F'))
```

```python
numbers = [1,2,3,4,5]

squares = list(map(lambda number:number**2, numbers))

print(squares)
```

```python
number_lst = [2,5,6,7,8,10,12,15,14]
def even_number(number):
    return number%2==0
even_num_lst = list(filter(even_number,number_lst))
print(even_num_lst)
```
