# 4 Recursion

## 4-1 Factorial using recursion 

```python
def fact(n):
    if(n==1 or n==0):
        return 1
    recval = fact(n-1)
    result = n*recval
    return result

n = int(input("Factorial - Enter number"))

print(f"Factorial {fact(n)}")
    
```

### 18 POWER 2 to N


```python
def power2n(n):
    if(n==1):
        return 2
    if(n==0):
        return 1

    recval = power2n(n-1)
    res = 2*recval
    return res;

n = int(input("Power2n - Enter number"))

print(f"Power2n {power2n(n)}")
```

## 4-2 SUM OF N NATURAL NUMBER

```python
def sum_of_number(n):
    if(n==1):
        return 1
    if(n==0):
        return 0

    recval = sum_of_number(n-1)

    result = n+recval
    return result

n = int(input("Enter number"))

print(f"Sum = {sum_of_number(n)}")
```
