# 2 Searching

## 2-1 Linear Searching 

```python
lst_numbers = [10,20,30,50,5,100]

def linear_search(arr,target):
    size = len(arr)
    for index in range(0, size):
        if(arr[index]==target):
            return index
    return -1

target = 50

result = linear_search(lst_numbers, target)
print(result)
```


## 2-2 Binary Search  

### list/array must be sorted 
### -1 means element not found 

```python
def binary_searching(arr, target):
    size = len(arr)
    start = 0
    end = size-1
    while (start<=end):
        mid = (start+end)//2
        print(f"mid {mid}")
        if(arr[mid]==target):
            return mid
        elif(arr[mid]>target):
            end = mid-1
        elif(arr[mid]<target):
            start = mid+1
        print(f"start {start} end {end}")

    return -1;
            


marks = [30,50,60,70,75,80,85,90,95]
print("Element found at index",binary_searching(marks,95))
```
