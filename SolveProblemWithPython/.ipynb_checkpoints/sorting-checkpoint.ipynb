### Sorting Algorithm

---
Bubble Sort
---

Bubble Sort is a simple, comparison-based sorting algorithm that works by repeatedly stepping through a list, comparing adjacent elements, and swapping them if they are in the wrong order. This process repeats until the entire list is completely sorted. It is nicknamed "bubble sort" because larger elements progressively "bubble up" to the end of the collection with each iteration

```python
def bubble_sort(arr):
    size_list = len(arr);
    for p in range(0,size_list):
        print(f"pass {p}")
        for j in range(size_list-p-1):
            if(arr[j]>arr[j+1]):
                temp = arr[j+1]
                arr[j+1]=arr[j]
                arr[j]= temp

    return arr;
                
                
    
    

lst_numbers = [12,11,80,9.5,30,40,20,75]

print("Sorted list",bubble_sort(lst_numbers))
```

---
Selection Sort 
--- 
Selection Sort is a simple, comparison-based sorting algorithm that works by repeatedly finding the smallest (minimum) element from the unsorted part of a list and moving it to the beginning

```python
def selection_sort(arr):
    size_list = len(arr);
    for p in range(0,size_list-1):
        print(f"pass {p}")
        min_index = p
        for j in range(p+1,size_list):
            if(arr[j]<arr[min_index]):
                min_index = j
        if(p!=min_index):
            print(f"min {min_index}")
            arr[p],arr[min_index] = arr[min_index],arr[p]

    return arr;
                
listA = [12,11,80,9.5,30,40,20,75]

print("Sorted list",selection_sort(listA))
```
