## 6-1 Delete a node - head 

```python
def deleteAtHead(head):
    if head is None:
        return None
    return head.next
values1 = input("Enter the values for the linked list (space-separated): ").split()
values2 = [int(x) for x in values1]

headn = create_ll(values2)
print("Original linked list:")
print_ll(headn)
head1 = deleteAtHead(headn)
print("Linked list after deleting head:")
print_ll(head1)
```
