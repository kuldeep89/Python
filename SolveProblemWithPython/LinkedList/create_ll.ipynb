# 5 Create linked list, Print

## 5-1 create , print

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

    def print_ll(head):
        first = head
        while(head.next!=None):
            print(head.data)
            head = head.next

first = Node(1)
second = Node(2)
third = Node(3)

first.next = second
second.next = third 

head = first
del(first)
del(second)
del(third)

def print_ll(head):
    temp = head
    while(temp!=None):
        print(temp.data, end="->")
        temp = temp.next
    return

print_ll(head)



    
```

## 5-2 create linked list with input from user

```python
def take_input():
    value = int(input("Enter the value of Node :-"))
    head = None
    while(value!=-1):
        newNode = Node(value)
        if(head==None):
            head = newNode
        else:
            print(newNode.data)
            temp = head;
            while(temp.next!=None):
                temp = temp.next
            temp.next = newNode

        value = int(input("Enter the value of Node :-"))
    return head

newHead = take_input()
print_ll(newHead)
```

## 5-3 create linked list with input from user --OPTIMIZED 

```python
def take_input():
    value = int(input("Enter the value of Node :-"))
    head = None
    tail = None
    while(value!=-1):
        newNode = Node(value)
        if(head==None):
            head = newNode
            tail = newNode
        else:
            tail.next = newNode;
            tail  = tail.next

        value = int(input("Enter the value of Node :-"))
    return head

newHead = take_input()
print_ll(newHead)
```

## 5-4 length of linked list

```python
def length_ll(head):
    temp = head
    length = 0
    while(temp!=None):
        print(temp.data, end="->")
        length = length+1
        temp = temp.next
    return length

length_of_ll = length_ll(newHead)
print(f"Length of LL - {length_of_ll}")
```

## 5-5 length of linked  --recursively

```python
def LLLengthRecursively(head):
    if(head==None):
        return 0
    ans = LLLengthRecursively(head.next)
    return ans + 1

newHead1 = take_input()
print_ll(newHead1)
length_ll = LLLengthRecursively(newHead1)
print(f"Length of LL - {length_ll}")
```

```python
def insertAtIndex(head,data,index):
    if(index==0):
        return insertAtHead(head,data)
    temp = head
    count = 0
    newNode = Node(data)
    while temp is not None and count < index-1 :
        temp = temp.next
        count = count+1
    newNode.next = temp.next
    temp.next = newNode
    return head

head4 = take_input()
newHead4 = insertAtIndex(head4, 20,3)
print_ll(newHead4)
```
