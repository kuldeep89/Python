# Pattern Problems


## 1 -1 Square of side N 



```python
number = int(input('Enter sides need in square'))

def generate_square(number):
    
    square_pattern = []
    for i in range(number):
        square_pattern.append("x"*number)
        print("x "*number)
    return square_pattern;

pattern = generate_square(number)
print(pattern)
```

## 1-2 Hollow Square Problem 

```python
hollow_square_sides = int(input('Enter sides need in square'))

def generate_hollow_square(number):
    
    pattern = []
    for i in range(number):
        if(i==0 or i==number-1):
            print("*"*number)
            pattern.append("*"*number)
        else:
            str_pattern = "*"
            str_pattern += " "*(number-2)
            str_pattern += "*"
            pattern.append(str_pattern)
            print(str_pattern)
    return pattern;

hollow_pattern = generate_hollow_square(hollow_square_sides)
print(hollow_pattern)
```

## 1-3 Rectangular Pattern

```python
def rectangular_pattern(rlen,rbreadth):
    rpattern = []
    for i in range(0,rlen):
        rpattern.append("x"*rbreadth)
        print("x"*rbreadth)
    return rpattern;
    


rect_length = int(input("Enter length "))
rect_breadth = int(input("Enter breadth"))
rpattern = rectangular_pattern(rect_length,rect_breadth)
print(rpattern)
```


## 1 - 4 Right Angled Triangle


```python
def right_angled_triangle_pattern(n):
    rpattern = [];
    for i in range(n):
        print("* "*(i+1));
        rpattern.append("* "*(i+1))
    return rpattern;

number_of_rows = int(input("Enter number of rows required "))
rag_pattern = right_angled_triangle_pattern(number_of_rows)
print(rag_pattern)
```

## 1-5 Inverted Right Angled Triangle

```python
def inverted_right_angled_triangle_pattern(n):
    ipattern = [];
    for i in range(n,0,-1):
        print("* "*(i));
        ipattern.append("* "*(i))
    return ipattern;

number_of_rows = int(input("Enter number of rows required "))
irag_pattern = inverted_right_angled_triangle_pattern(number_of_rows)
print(irag_pattern)
```

## 1- 6 Pyramid with stars


```python
def pyramid_pattern(n):
    ppattern = [];
    for i in range(1,n+1):
        print(" "*(n-i)+"*"*(2*i-1)+" "*(n-i));
        ppattern.append(" "*(n-i)+"X"*(2*i-1)+" "*(n-i))
    return ppattern;

number_of_rows = int(input("Enter number of rows required "))
ppattern = pyramid_pattern(number_of_rows)
print(ppattern)
    
```

## 1-7 Inverted Pyramid with stars

```python
def inverted_pyramid_pattern(n):
    ppattern = [];
    for i in range(1,n+1):
        print(" "*(i-1)+"X"*(2*n-(2*i-1))+" "*(i-1));
        ppattern.append(" "*(i-1)+"*"*(2*n-(2*i-1))+" "*(i-1))
    return ppattern;

number_of_rows = int(input("Enter number of rows required "))
ippattern = inverted_pyramid_pattern(number_of_rows)
print(ippattern)
    
```

## 1-8 Right Angled Triangle with Number




```python
def right_angled_triangle_with_number_pattern(n):
    rpattern = [];
    for i in range(n):
        print(f"{i+1}"*(i+1));
        rpattern.append(f"{i+1}"*(i+1))
    return rpattern;

number_of_rows = int(input("Enter number of rows required "))
rag_pattern = right_angled_triangle_with_number_pattern(number_of_rows)
print(rag_pattern)
```

## 1-9 Floyds Triangle 

```python
def floyds_triangle_pattern(n):
    fpattern = [];
    c = 1; #tracker printed number last row 
    for i in range(n):
        print_number = ""
        for j in range(i+1):
            print_number +=f"{c}"
            c+=1    
        print(f"{print_number}");
        fpattern.append(f"{print_number}")
    return fpattern;

number_of_rows = int(input("Enter number of rows required "))
floyds_pattern = floyds_triangle_pattern(number_of_rows)
print(floyds_pattern)
```

## 1-10 Diamond Pattern

def diamond_pattern(n): 
    dpattern = []
    #Top half
    for i in range(0,n+1):
        print(" "*(n-i),"*"*(2*i-1)," "*(n-i))
        dpattern.append(" "*(n-i)+"*"*(2*i-1)+" "*(n-i))

    #Bottom half 
    for j in range (n-1,0,-1):
        print(" "*(n-j),"*"*(2*j-1)," "*(n-j))
        dpattern.append(" "*(n-j)+"*"*(2*j-1)+" "*(n-j))
    return dpattern;
    

number_of_rows = int(input("Enter number of rows required "))
diamond_pattern = diamond_pattern(number_of_rows)
print(diamond_pattern)

## 1-11 Right Angled Triangle

```python
def rat_pattern(n): 
    pattern = []
    #Top half
    for i in range(0,n):
        print(" "*(n-i-1),"X"*(i+1))
        pattern.append(" "*(n-i-1)+"X"*(i+1))

    return pattern;
    

number_of_rows = int(input("Enter number of rows required "))
pattern = rat_pattern(number_of_rows)
print(pattern)
```
